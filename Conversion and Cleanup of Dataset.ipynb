# **Converting all Images to a Consistent Format (e.g., RGB .jpg)**

This ensures:

* The image is a valid JPEG
* It uses RGB (3 channels), which TensorFlow expects
* Any format issues (e.g., .png with alpha, .bmp, or weird headers) are corrected
* Removes subtle corruption or unsupported formats

```python
from PIL import Image
import os

data_dir = 'D:\\Accelx\\Dataset'

for subdir, dirs, files in os.walk(data_dir):
    for file in files:
        file_path = os.path.join(subdir, file)
        try:
            with Image.open(file_path) as img:
                # Convert to RGB and save as JPG to ensure consistency
                img = img.convert("RGB")
                img.save(file_path, format='JPEG')
        except Exception as e:
            print(f"Failed to convert {file_path}: {e}")
            os.remove(file_path)
```
